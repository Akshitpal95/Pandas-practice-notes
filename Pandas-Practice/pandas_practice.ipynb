```python
#loc = label, iloc = integer position.

data = {
    "Name": ["Aman", "Rahul", "Priya"],
    "Age": [19, 20, 18],
    "Marks": [85, 90, 95]
}

df = pd.DataFrame(data)

print(df.loc[0])
```

```python
print(df.loc[1, "Name"]) 
```

```python
print(df.iloc[1, 0])
```

```python
df.loc[0:1, ["Name", "Marks"]]

df.iloc[0:2, 0:2]

df.loc[2, "Age"] 
```

### Conditional Filtering

```python
print(df[df["Age"] > 18])
```

```python
# Age greater than 18 AND marks above 85
df[(df["Age"] > 18) & (df["Marks"] > 85)]
```

```python
# Age equal to 18 OR marks above 90
df[(df["Age"] == 18) | (df["Marks"] > 90)]
```

```python
df[df["Name"].isin(["Aman", "Priya"])]
```

```python
# Sort by marks (ascending)
df.sort_values(by="Marks")

# Highest marks first
df.sort_values(by="Marks", ascending=False)
```

```python
data = {
    "Department": ["IT", "CS", "IT", "CS"],
    "Salary": [30000, 40000, 35000, 45000]
}

df2 = pd.DataFrame(data)

print(df2.groupby("Department")["Salary"].mean())
```

```python
df2.groupby("Department")["Salary"].sum()
df2.groupby("Department")["Salary"].max()
df2.groupby("Department")["Salary"].count()
```

```python
print(df2["Department"].value_counts())
```

```python
df2["Department"].value_counts(normalize=True) * 100
```

```python
# Find duplicate rows
df2.duplicated()

# Remove duplicate rows
df2.drop_duplicates()

# Remove duplicates based on one column
df2.drop_duplicates(subset=["Department"])
```

#### Quick revision
* Function &	Purpose
* sort_values()	Sort data
* groupby()	Group and calculate
* value_counts()	Count unique values
* drop_duplicates()	Remove repeated rows
